##### K-Nearest Neighbors (KNN)

1. Similar points lie close to each other, so nearby points share the same class.
2. Distance is meaningful, so features should be on a similar scale.
3. All features are equally important unless weighted.

##### Decision Tree

- Data can be separated by splitting one feature at a time (axis parallel splits).
- Each split is chosen greedily, keeping the locally best split.
- Observations are independent and the training data represents the real data.
- No assumption about the data distribution or linearity, and scaling is not needed.
- Without depth limits or pruning, the tree tends to overfit.

##### Random Forest

- Built on the decision tree assumptions.
- Trees are trained on random samples of data and random subsets of features to keep them uncorrelated.
- Observations are independent and the training data represents the real data.
- No distribution or scaling assumptions.
