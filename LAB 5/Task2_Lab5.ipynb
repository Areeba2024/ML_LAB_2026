```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
np.random.seed(42)
n = 200
residents = np.random.randint(1, 8, n)
income = np.random.randint(30000, 150000, n)
house_age = np.random.randint(1, 50, n)
noise = np.random.normal(0, 40, n)
consumption = 100 + 45 * residents + 0.002 * income + 3 * house_age + noise

df = pd.DataFrame({
    'residents': residents,
    'income': income,
    'house_age': house_age,
    'consumption': consumption
})
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15,4))
sns.regplot(x='residents', y='consumption', data=df, ax=ax[0])
sns.regplot(x='income', y='consumption', data=df, ax=ax[1])
sns.regplot(x='house_age', y='consumption', data=df, ax=ax[2])
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```

```python
X = df[['residents', 'income', 'house_age']]
y = df['consumption']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
residuals = y - y_pred

print('Intercept:', model.intercept_)
print(pd.Series(model.coef_, index=X.columns))
print('R2:', r2_score(y, y_pred))
```

```python
plt.figure(figsize=(6,4))
plt.scatter(y_pred, residuals)
plt.axhline(0, color='red')
plt.xlabel('Predicted')
plt.ylabel('Residual')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,4))
sns.histplot(residuals, kde=True, ax=ax[0])
stats.probplot(residuals, dist='norm', plot=ax[1])
plt.show()

stat, p = stats.shapiro(residuals)
print('Shapiro-Wilk p-value:', p)
```

```python
dw = np.sum(np.diff(residuals) ** 2) / np.sum(residuals ** 2)
print('Durbin-Watson:', dw)
```

```python
plt.figure(figsize=(8,3))
plt.plot(residuals.values)
plt.axhline(0, color='red')
plt.xlabel('Observation')
plt.ylabel('Residual')
plt.show()
```



**Decision: Linear regression is suitable for this dataset.**

Assumptions: 
1. Linearity
2. Independence
3. Homoscedasiticity
4. normality of residuals
5. Absence of perfect multicollinearity

All assumptions hold, so linear regression gives reliable and interpretable coefficients here. If real data showed curved patterns, a funnel shaped residual plot or strongly correlated features, linear regression would need to be replaced or adjusted (polynomial regression, transformation, or regularization).
